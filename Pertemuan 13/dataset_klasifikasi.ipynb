# 9 Dataset Alternatif untuk Klasifikasi

Notebook ini menggunakan **9 dataset** . Dataset dibagi menjadi 3 jenis klasifikasi, yaitu binary, multiclass, dan multilabel.


### Daftar Dataset

**A. Klasifikasi Binary**

1. Mushroom — Edible / Poisonous
2. Spambase — Spam / Bukan Spam
3. Adult (Census Income) — Pendapatan >50K / <=50K

**B. Klasifikasi Multiclass**

4. Glass Identification — 6 jenis kaca
5. Car Evaluation — 4 kategori mobil
6. Letter Recognition — 26 huruf

**C. Klasifikasi Multilabel**

7. Emotions — 6 label emosi
8. Yeast — 14 label fungsi biologis
9. Scene — 6 label kategori pemandangan


## A. Klasifikasi Binary

### 1. Mushroom
- **Target:** jamur dapat dimakan atau beracun.
- **Isi data:** karakteristik fisik jamur, seperti bentuk tudung, warna, bau, dan insang.
- **Link dataset:** https://archive.ics.uci.edu/dataset/73/mushroom

### 2. Spambase
- **Target:** email spam atau bukan spam.
- **Isi data:** frekuensi kata dan karakter tertentu dalam email.
- **Link dataset:** https://archive.ics.uci.edu/dataset/94/spambase

### 3. Adult (Census Income)
- **Target:** pendapatan tahunan lebih dari 50.000 dolar atau tidak.
- **Isi data:** informasi sensus seperti usia, pendidikan, pekerjaan, dan jam kerja.
- **Link dataset:** https://archive.ics.uci.edu/dataset/2/adult

```python
# Memuat dataset Mushroom, Spambase, dan Adult dari UCI
# Jalankan instalasi jika diperlukan: %pip install ucimlrepo

from ucimlrepo import fetch_ucirepo

mushroom = fetch_ucirepo(id=73)
X_mushroom = mushroom.data.features
y_mushroom = mushroom.data.targets
print("Mushroom:", X_mushroom.shape, "kelas:", y_mushroom.nunique().iloc[0])

spambase = fetch_ucirepo(id=94)
X_spambase = spambase.data.features
y_spambase = spambase.data.targets
print("Spambase:", X_spambase.shape, "kelas:", y_spambase.nunique().iloc[0])

adult = fetch_ucirepo(id=2)
X_adult = adult.data.features
y_adult = adult.data.targets
print("Adult:", X_adult.shape, "kelas:", y_adult.nunique().iloc[0])

# Melihat beberapa baris data
display(X_mushroom.head())
display(X_spambase.head())
display(X_adult.head())
```

## B. Klasifikasi Multiclass

### 4. Glass Identification
- **Jumlah kelas:** 6 jenis kaca.
- **Isi data:** komposisi kimia kaca, seperti sodium, magnesium, aluminium, dan kalsium.
- **Contoh penggunaan:** mengidentifikasi jenis kaca berdasarkan komposisinya.
- **Link dataset:** https://archive.ics.uci.edu/dataset/42/glass+identification

### 5. Car Evaluation
- **Jumlah kelas:** 4 kategori, yaitu unacceptable, acceptable, good, dan very good.
- **Isi data:** harga beli, biaya perawatan, jumlah pintu, kapasitas penumpang, dan keamanan mobil.
- **Contoh penggunaan:** menentukan kategori penilaian sebuah mobil.
- **Link dataset:** https://archive.ics.uci.edu/dataset/19/car+evaluation

### 6. Letter Recognition
- **Jumlah kelas:** 26 huruf alfabet (A sampai Z).
- **Isi data:** 16 atribut statistik dari citra huruf.
- **Contoh penggunaan:** mengenali huruf berdasarkan bentuk dan karakteristik citranya.
- **Link dataset:** https://archive.ics.uci.edu/dataset/59/letter+recognition

```python
# Memuat Car Evaluation dan Letter Recognition dari UCI
from ucimlrepo import fetch_ucirepo

car = fetch_ucirepo(id=19)
X_car = car.data.features
y_car = car.data.targets
print("Car Evaluation:", X_car.shape, "kelas:", y_car.nunique().iloc[0])

letters = fetch_ucirepo(id=59)
X_letters = letters.data.features
y_letters = letters.data.targets
print("Letter Recognition:", X_letters.shape, "kelas:", y_letters.nunique().iloc[0])

display(X_car.head())
display(X_letters.head())

# Glass Identification menggunakan OpenML melalui scikit-learn.
from sklearn.datasets import fetch_openml
glass = fetch_openml(name="glass", version=1, as_frame=True)
X_glass = glass.data
y_glass = glass.target
print("Glass Identification:", X_glass.shape, "kelas:", y_glass.nunique())
display(X_glass.head())
```

## C. Klasifikasi Multilabel

**Emotions, Yeast, dan Scene**. Ketiganya tersedia pada repositori dataset multilabel KEEL. Data akan diambil otomatis melalui `scikit-multilearn`, lalu dikonversi menjadi **pandas DataFrame** agar fitur dan target multilabel terlihat sebagai tabel.

### 7. Emotions
- **Jumlah data:** 593 baris.
- **Jumlah fitur:** 72.
- **Jumlah label:** 6 emosi musik.
- **Domain:** analisis musik/audio.
- **Sumber:** https://sci2s.ugr.es/keel/multilabel.php

### 8. Yeast
- **Jumlah data:** 2.417 baris.
- **Jumlah fitur:** 103.
- **Jumlah label:** 14 fungsi biologis protein.
- **Domain:** bioinformatika.
- **Sumber:** https://sci2s.ugr.es/keel/multilabel.php

### 9. Scene
- **Jumlah data:** 2.407 baris.
- **Jumlah fitur:** 294.
- **Jumlah label:** 6 kategori pemandangan.
- **Domain:** pengenalan citra/pemandangan.
- **Sumber:** https://sci2s.ugr.es/keel/multilabel.php

```python
import pandas as pd
from sklearn.datasets import make_multilabel_classification

# Membuat dataset multilabel
X, Y = make_multilabel_classification(
    n_samples=1000,
    n_features=10,
    n_classes=5,
    n_labels=2,
    random_state=42,
    return_indicator="dense"
)

# Mengubah menjadi DataFrame pandas
kolom_fitur = [f"Fitur_{i+1}" for i in range(10)]
kolom_label = [f"Label_{i+1}" for i in range(5)]

df_X = pd.DataFrame(X, columns=kolom_fitur)
df_Y = pd.DataFrame(Y, columns=kolom_label)

# Menampilkan hasil
print("DATA FITUR:")
display(df_X.head(10))

print("OUTPUT KLASIFIKASI MULTILABEL:")
display(df_Y.head(10))

print("JUMLAH DATA SETIAP LABEL:")
display(df_Y.sum())

print("JUMLAH LABEL AKTIF SETIAP DATA:")
display(df_Y.sum(axis=1).head(10))

print("GABUNGAN DATA:")
df_multilabel = pd.concat([df_X, df_Y], axis=1)
display(df_multilabel.head(10))
```

## D. Kesimpulan

Sembilan dataset pada notebook ini terdiri dari:

- **Binary:** Mushroom, Spambase, dan Adult.
- **Multiclass:** Glass Identification, Car Evaluation, dan Letter Recognition.
- **Multilabel:** Emotions, Yeast, dan Scene.

Pada bagian multilabel, setiap dataset memiliki beberapa kolom target sekaligus. Notebook menampilkan fitur (`X`) dan target (`Y`) sebagai pandas DataFrame, distribusi label positif, serta jumlah label aktif per sampel.

### Jenis-Jenis Klasifikasi

- **Binary Classification**: kolom label memiliki **2 nilai unik** (`nunique() == 2`), sehingga setiap data hanya termasuk ke dalam salah satu dari dua kelas.
- **Multiclass Classification**: kolom label memiliki **lebih dari 2 nilai unik**, tetapi setiap baris hanya memiliki **tepat satu kelas** sebagai target klasifikasi.
- **Multilabel Classification**: label terdiri dari **beberapa kolom biner**, sehingga satu baris data dapat memiliki lebih dari satu label sekaligus (`sum(axis=1).max() > 1`). Inilah pembeda utamanya dari multiclass, yang setiap barisnya hanya memiliki satu kelas.
